# Selenium

**Realiza os imports necessários**

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options # Utilizaremos o navegador Chrome
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
```

**Especifica a página desejada**

```python
URL = 'https://quotes.toscrape.com/'
URL2 = 'https://python.org'
```

**Configura opções para o Driver**

```python
options = Options()

# Habilita o navegador headless (não abre a janela)
# options.add_argument('--headless=new')

# Define o tamanho da tela  
# options.add_argument("--window-size=1920,1080")

# Já abre o navegador em tela cheia
# options.add_argument("--start-maximized")

# Abre no modo anônimo
options.add_argument('--incognito')

# Define o idioma padrão do navegador 
options.add_argument('--lang=pt-BR')

# Desativa a barra superior "O Chrome está sendo controlado por um software de teste"
options.add_experimental_option("excludeSwitches", ["enable-automation"])
```

**Inicializa o navegador**

```python
driver = webdriver.Chrome(options=options)
```

**Acessa a página desejada**

```python
driver.get(URL)
```

## Controles de Navegação

**Acessa uma nova página**

```python
driver.get(URL2)
```

**Volta para a página anterior**

```python
driver.back()
```

**Avança para a nova página**

```python
driver.forward()
```

**Recarrega a página**

```python
driver.refresh()
```

**Volta para a página inicial**

```python
driver.get(URL)
```

## Localização de Elementos

Para extrir uma informação ou clicar em um botão, o Selenium precisa apenas encontrar o nó correspondente na árvore HTML da página.

**Buscando o título pela tag HTML (TAG_NAME)**

```python
# find_element() -> retorna apenas o primeiro elemento encontrado
titulo_pagina = driver.find_element(By.TAG_NAME, 'h1')
print(f'Título da página: {titulo_pagina.text}')
```

**Conta quantas citações existem na página utilizando o nome da classe CSS (CSS_SELECTOR)**

```python
# find_elements() -> retorna todos os elementos encontrados
citacoes = driver.find_elements(By.CLASS_NAME, 'quote')
print(f'Número de citações na página: {len(citacoes)}')
```

## Interação com a Página

```python
# Acessa a página de login
driver.get(URL + '/login')
```

**Localiza os campos de usuário e senha**

```python
campo_usuario = driver.find_element(By.ID, 'username')
campo_senha = driver.find_element(By.ID, 'password')
```

**Insere usuário e senha quaisquer**

```python
campo_usuario.send_keys("usuario123")
campo_senha.send_keys("senha_forte123")
```

**Confirma usando a tecla Enter no campo senha**

```python
campo_senha.send_keys(Keys.ENTER)
```

**Fecha o navegador para não deixar processos mortos na RAM**
