### Agrupa o Total de Votos por Candidato em Ambos os Turnos 

Agrupamentos feitos por **Local de Votação**, **Seções** e **Bairros**

**Agrupa por Locais de Votação**

```python
# Seleciona o nome do local com mais eleitores para cada par [lat, lon]
nm_local_principal_1T = (
    df_turno1.groupby(['NR_LATITUDE', 'NR_LONGITUDE', 'NM_LOCAL_VOTACAO'], as_index=False)['QT_VOTOS']
        .sum()
        .sort_values(by='QT_VOTOS', ascending=False)
        .drop_duplicates(subset=['NR_LATITUDE', 'NR_LONGITUDE'])
        [['NR_LATITUDE', 'NR_LONGITUDE', 'NM_LOCAL_VOTACAO']]
)

# Agrupa os votos por [lat, lon] e candidato, e traz o nome do local de votacao obtido anteriormente
df_locais_1T = (
    df_turno1.groupby(['NR_LATITUDE', 'NR_LONGITUDE', 'NM_VOTAVEL', "NR_VOTAVEL"], as_index=False)['QT_VOTOS']
        .sum()
        .merge(nm_local_principal_1T, on=['NR_LATITUDE', 'NR_LONGITUDE'], how='left')
        .sort_values(by=['NM_LOCAL_VOTACAO', 'QT_VOTOS'], ascending=[True, False],)
)

# Calcula o percentual de votos de cada candidato por local 
total_votos_local1T = df_locais_1T.groupby(['NR_LATITUDE', 'NR_LONGITUDE'])['QT_VOTOS'].transform('sum')
df_locais_1T['PCT_VOTOS_LOCAL'] = (df_locais_1T['QT_VOTOS'] / total_votos_local1T) * 100.00
df_locais_1T['PCT_VOTOS_LOCAL'] = df_locais_1T['PCT_VOTOS_LOCAL'].round(2)

# Calcula também o percentual de votos válidos de cada candidato por local
votos_invalidos = ['Branco', 'Nulo']
votos_validos1T = df_locais_1T['QT_VOTOS'].where(~df_locais_1T['NM_VOTAVEL'].isin(votos_invalidos), 0)
total_votos_validos1T = votos_validos1T.groupby([df_locais_1T['NR_LATITUDE'], df_locais_1T['NR_LONGITUDE']]).transform('sum')
df_locais_1T['PCT_VOTOS_VALIDOS_LOCAL'] = (votos_validos1T / total_votos_validos1T) * 100.00
df_locais_1T['PCT_VOTOS_VALIDOS_LOCAL'] = df_locais_1T['PCT_VOTOS_VALIDOS_LOCAL'].round(2)

# Adiciona o ganhador de cada local
mais_votado_local = df_locais_1T.groupby(['NR_LATITUDE', 'NR_LONGITUDE'])['QT_VOTOS'].transform('max')
df_locais_1T['MAIS_VOTADO_LOCAL'] = (df_locais_1T['QT_VOTOS'] == mais_votado_local)

df_locais_1T.head(10)
```

```python
# Seleciona o nome do local com mais eleitores para cada par [lat, lon]
nm_local_principal_2T = (
    df_turno2.groupby(['NR_LATITUDE', 'NR_LONGITUDE', 'NM_LOCAL_VOTACAO'], as_index=False)['QT_VOTOS']
        .sum()
        .sort_values(by='QT_VOTOS', ascending=False)
        .drop_duplicates(subset=['NR_LATITUDE', 'NR_LONGITUDE'])
        [['NR_LATITUDE', 'NR_LONGITUDE', 'NM_LOCAL_VOTACAO']]
)

# Agrupa os votos por [lat, lon] e candidato, e traz o nome do local de votacao obtido anteriormente
df_locais_2T = (
    df_turno2.groupby(['NR_LATITUDE', 'NR_LONGITUDE', 'NM_VOTAVEL', "NR_VOTAVEL"], as_index=False)['QT_VOTOS']
        .sum()
        .merge(nm_local_principal_2T, on=['NR_LATITUDE', 'NR_LONGITUDE'], how='left')
        .sort_values(by=['NM_LOCAL_VOTACAO', 'QT_VOTOS'], ascending=[True, False],)
)

# Calcula o percentual de votos de cada candidato por local 
total_votos_local2T = df_locais_2T.groupby(['NR_LATITUDE', 'NR_LONGITUDE'])['QT_VOTOS'].transform('sum')
df_locais_2T['PCT_VOTOS_LOCAL'] = (df_locais_2T['QT_VOTOS'] / total_votos_local2T) * 100.00
df_locais_2T['PCT_VOTOS_LOCAL'] = df_locais_2T['PCT_VOTOS_LOCAL'].round(2)

# Calcula também o percentual de votos válidos de cada candidato por local
votos_invalidos = ['Branco', 'Nulo']
votos_validos2T = df_locais_2T['QT_VOTOS'].where(~df_locais_2T['NM_VOTAVEL'].isin(votos_invalidos), 0)
total_votos_validos2T = votos_validos2T.groupby([df_locais_2T['NR_LATITUDE'], df_locais_2T['NR_LONGITUDE']]).transform('sum')
df_locais_2T['PCT_VOTOS_VALIDOS_LOCAL'] = (votos_validos2T / total_votos_validos2T) * 100.00
df_locais_2T['PCT_VOTOS_VALIDOS_LOCAL'] = df_locais_2T['PCT_VOTOS_VALIDOS_LOCAL'].round(2)

# Adiciona o ganhador de cada local
mais_votado_local = df_locais_2T.groupby(['NR_LATITUDE', 'NR_LONGITUDE'])['QT_VOTOS'].transform('max')
df_locais_2T['MAIS_VOTADO_LOCAL'] = (df_locais_2T['QT_VOTOS'] == mais_votado_local)

df_locais_2T.head(10)
```

**Calcula a "força" do candidato por local em relação às demais votacoes pela cidade**

```python
# Usa os percentis 10% e 90% para mitigar o impacto dos outliers 
p10_votos_candidato = df_locais_1T.groupby('NR_VOTAVEL')['PCT_VOTOS_LOCAL'].transform(lambda x: x.quantile(0.10))
p90_votos_candidato = df_locais_1T.groupby('NR_VOTAVEL')['PCT_VOTOS_LOCAL'].transform(lambda x: x.quantile(0.90))

normaliza_votos = ((df_locais_1T['PCT_VOTOS_LOCAL'] - p10_votos_candidato) / (p90_votos_candidato - p10_votos_candidato)).clip(0, 1)
df_locais_1T['DOMINANCIA_LOCAL'] = normaliza_votos.round(2)
```

```python
# Usa os percentis 10% e 90% para mitigar o impacto dos outliers 
p10_votos_candidato = df_locais_2T.groupby('NR_VOTAVEL')['PCT_VOTOS_LOCAL'].transform(lambda x: x.quantile(0.10))
p90_votos_candidato = df_locais_2T.groupby('NR_VOTAVEL')['PCT_VOTOS_LOCAL'].transform(lambda x: x.quantile(0.90))

normaliza_votos = ((df_locais_2T['PCT_VOTOS_LOCAL'] - p10_votos_candidato) / (p90_votos_candidato - p10_votos_candidato)).clip(0, 1)
df_locais_2T['DOMINANCIA_LOCAL'] = normaliza_votos.round(2)
```

## Choropleth Maps do 1º Turno

**Lista os candidatos e as cores atribuidas para cada um deles**

```python
opcoes_voto = [
    "Lula", "Renan Santos", "Hertz Dias", "Edmilson Costa", "Flavio Bolsonaro", "Clariana Barao", "Leonardo Avalanche", "Rui Costa Pimenta", "Zema", "Veterinário Wilson Grassi", "Ronaldo Caiado", "Escritor Augusto Cury", "Samara", "Branco", "Nulo",
]

cores_partidos = {
    13: "#ff0000", # Lula
    14: "#f0d01a", # Renan Santos
    16: "#780f07", # Hertz Dias
    21: "#fc4521", # Edmilson Costa
    22: "#0c7a0a", # Flavio Bolsonaro
    27: "#80d7ff", # Clariana Barão 
    28: "#0a1e57", # Leonardo Avalanche
    29: "#9c2308", # Rui Costa Pimenta  
    30: "#e66909", # Zema
    35: "#187031", # Veterinário Wilson Grassi
    55: "#0306a6", # Ronaldo Caiado
    70: "#077375", # Escritor Augusto Cury
    80: "#000000", # Samara Martins
    95: "#cfcfcf", # Branco 
    96: "#a3a3a3", # Nulo
}
```

**Função para atribuir paleta de cores para cada candidato**

```python
def candidate_color_func(party:int):
    if party in cores_partidos.keys():
        return cores_partidos[party]
    
    # Outros? (Nao é pra ter)
    return "#ff00dd"
```

**Cria o mapa em SJDR**

```python
sjdr_lat_lon = [-21.1365, -44.2619]

mapa_1T_locais = folium.Map(
    location=sjdr_lat_lon,
    zoom_start=14,
    tiles='https://tile.openstreetmap.de/{z}/{x}/{y}.png',
    attr='© OpenStreetMap contributors',
    control_scale=True, # Mostra escala do mapa no canto inferior esquerdo
    
    # Não permite sair da região do mapa
    max_bounds=True, 
    min_lat=sjdr_lat_lon[0] + 1.0,
    max_lat=sjdr_lat_lon[0] - 1.0,
    min_lon=sjdr_lat_lon[1] + 1.0,
    max_lon=sjdr_lat_lon[1] - 1.0,
)

# Plugin de deixar o mapa em tela cheia
Fullscreen(
    position="topright",
    force_separate_button=True,
    title="Entrar no Modo Tela Cheia",
    title_cancel="Sair do Modo Tela Cheia",
).add_to(mapa_1T_locais)
```

**Cria uma camada para cada uma das opcoes de voto**

```python
camadas_candidatos_1T = {
    opcao_voto: folium.FeatureGroup(
        name=opcao_voto,
        show=(opcao_voto == "Lula") # Deixa o Lula por padrao 
    )
    for opcao_voto in opcoes_voto
}
```

**Coloca os marcadores de votos em cada um dos locais de votacao, para cada uma das opcoes de voto**

```python
def get_secoes(secoes:list):
    secoes_str = '' 
    for sec in range(len(secoes) - 1):
        secoes_str += f'{secoes[sec]}, ' 
    secoes_str += str(secoes[len(secoes)-1])
    return secoes_str

estilo_tooltip = """
    background-color: #ffffff;
    color: #000000;
    font-family: 'Helvetica Neue', Arial, sans-serif;
    font-size: 13px;
    border-radius: 6px;
    border: 1px solid #444444;
    padding: 8px 12px;
    box-shadow: 3px 3px 10px rgba(0, 0, 0, 0.4);
"""

for row in df_locais_1T.itertuples(index=False):
    lat = row.NR_LATITUDE 
    lon = row.NR_LONGITUDE 

    candidato = str(row.NM_VOTAVEL).title()
    if candidato not in camadas_candidatos_1T:
        print(f'ERRO! {candidato} não está entre os candidatos listados.')
        continue
    camada = camadas_candidatos_1T[candidato] 
    
    pct = row.PCT_VOTOS_LOCAL
    party = row.NR_VOTAVEL
    color = candidate_color_func(party)
    votos = row.QT_VOTOS
    votos_validos = row.PCT_VOTOS_VALIDOS_LOCAL
   
    nm_local = row.NM_LOCAL_VOTACAO 
  
    vencedor = row.MAIS_VOTADO_LOCAL
   
    secoes = set(df_turno1.loc[(df_turno1['NR_LATITUDE'] == lat) & (df_turno1['NR_LONGITUDE'] == lon), 'NR_SECAO'].to_list())

    html_popup = str()
    if vencedor: 
        html_popup = f'<h3 style="color: {color};"><b>{candidato}<span style="color: #fca503;"> (Vencedor)</b></span></h3><br>' \
                     f"<b>Local:</b> {nm_local}<br>" \
                     f'<b>Votos:</b> {votos} ({pct}%)<br>' \
                     f'<b>Votos Validos:</b> {votos_validos}%<br>' \
                     f"<b>Seções:</b> {get_secoes(list(secoes))}"
    else:
        html_popup = f'<h3 style="color: {color};"><b>{candidato}</b></h3><br>' \
                     f"<b>Local:</b> {nm_local}<br>" \
                     f"<b>Votos:</b> {votos} ({pct}%)<br>" \
                     f"<b>Votos Validos:</b> {votos_validos}%<br>" \
                     f"<b>Seções:</b> {get_secoes(list(secoes))}"
    
    geojson_point = {
        "type": "FeatureCollection",
        "features": [
            {
                "type": "Feature",
                "geometry": {
                    "type": "Point",
                    "coordinates": [lon, lat]
                },
                "properties": {
                    "candidato": candidato,
                    "local": nm_local,
                    "votos": f"{votos} ({pct}%)",
                    "votos_validos": f"{votos_validos}%"
                }
            }
        ]
    }
    
    tooltip_customizado = folium.GeoJsonTooltip(
        fields=['local'],
        aliases=['Local:'],
        style=estilo_tooltip,
        sticky=True,
        localize=True
    )
    
    marker = folium.Circle(
        radius=math.sqrt(row.QT_VOTOS) * 5,
        color=color if not vencedor else '#fca503',
        fill=True,
        fill_color=color,
        fill_opacity=min(max(0.10, row.DOMINANCIA_LOCAL), 0.80)
    )
   
    popup = folium.Popup(
        html_popup,
        max_width=350
    )
    
    folium.GeoJson(
        geojson_point,
        marker=marker,
        popup=popup,
        tooltip=tooltip_customizado    
    ).add_to(camada)
   
# Adiciona as camadas ao mapa 
for camada in camadas_candidatos_1T.values():
    camada.add_to(mapa_1T_locais)
folium.LayerControl(collapsed=False).add_to(mapa_1T_locais)
```

**Justificativa escolhas gráficas**

Decidimos por utilizar, para cada candidato, a cor dominante do partido/campanha, para que facilite a associação do leitor da cor com o candidato em questão.

Também decidimos por utilizar uma opacidade diferente em cada local para representar a força da votação do candidato naquele local. A dominância é calculada em relação com as demais votações obtidas no município. Utilizamos os percentis 10% e 90% como votações de referência para não sofrer o impacto causado por outliers.

Neste mapa, realizamos uma visualização em camadas para que o leitor possa comparar facilmente qual dos candidatos teve votação mais expressiva naquele local. Também é possível que o leitor compare o impacto dos votos brancos/nulos em cada uma das seções, se ele se interessar. 

Quanto à escala, escolhemos uma que fosse capaz de mostrar todo o mapa de São João del-Rei já a primeira vista, para que o panorama prinicipal já ficasse a disposição do leitor a primeira vista. 

Por fim, utilizamos uma borda amarela/laranja para marcar que o candidato foi aquele com maior votação naquele local.

**Interpretação Mapa**

Com base nos fatos detalhados no mapa, podemos afirmar que o município de São João del-Rei apresenta padrões diferentes de votação em diferentes regiões da cidade. Na região leste da cidade, o espectro político da direita tem bem mais força que o da esquerda, padrão que se reverte no lado oeste da cidade.

Ambos os espectros políticos também apresentam locais de grande dominância nas suas regiões de influência, como o bairro Dom Bosco para a esquerda e o bairro Matosinhos para a direita.

Apesar de se tratar do 1º turno, os votos da cidade se dividiram majoritariamente nos dois principais candidatos do pleito, Flávio Bolsonaro, para a direita, e Lula, para a esquerda. 

Também é possível observar que a parcela de votos brancos e nulos neste turno foi bastante expressiva na cidade, podendo facilmente mudar o rumo do panorama eleitoral caso estes eleitores tivessem comparecido às urnas.

```python
def get_secoes(secoes:list):
    secoes_str = '' 
    for sec in range(len(secoes) - 1):
        secoes_str += f'{secoes[sec]}, ' 
    secoes_str += str(secoes[len(secoes)-1])
    return secoes_str

estilo_tooltip = """
    background-color: #ffffff;
    color: #000000;
    font-family: 'Helvetica Neue', Arial, sans-serif;
    font-size: 13px;
    border-radius: 6px;
    border: 1px solid #444444;
    padding: 8px 12px;
    box-shadow: 3px 3px 10px rgba(0, 0, 0, 0.4);
"""

for row in df_locais_2T.itertuples(index=False):
    lat = row.NR_LATITUDE 
    lon = row.NR_LONGITUDE 

    candidato = str(row.NM_VOTAVEL).title()
    if candidato not in camadas_candidatos_2T:
        print(f'ERRO! {candidato} não está entre os candidatos listados.')
        continue
    camada = camadas_candidatos_2T[candidato] 
    
    pct = row.PCT_VOTOS_LOCAL
    party = row.NR_VOTAVEL
    color = candidate_color_func(party)
    votos = row.QT_VOTOS
    votos_validos = row.PCT_VOTOS_VALIDOS_LOCAL
   
    nm_local = row.NM_LOCAL_VOTACAO 
  
    vencedor = row.MAIS_VOTADO_LOCAL
   
    secoes = set(df_turno2.loc[(df_turno2['NR_LATITUDE'] == lat) & (df_turno2['NR_LONGITUDE'] == lon), 'NR_SECAO'].to_list())

    html_popup = str()
    if vencedor: 
        html_popup = f'<h3 style="color: {color};"><b>{candidato}<span style="color: #fca503;"> (Vencedor)</b></span></h3><br>' \
                     f"<b>Local:</b> {nm_local}<br>" \
                     f'<b>Votos:</b> {votos} ({pct}%)<br>' \
                     f'<b>Votos Validos:</b> {votos_validos}%<br>' \
                     f"<b>Seções:</b> {get_secoes(list(secoes))}"
    else:
        html_popup = f'<h3 style="color: {color};"><b>{candidato}</b></h3><br>' \
                     f"<b>Local:</b> {nm_local}<br>" \
                     f"<b>Votos:</b> {votos} ({pct}%)<br>" \
                     f"<b>Votos Validos:</b> {votos_validos}%<br>" \
                     f"<b>Seções:</b> {get_secoes(list(secoes))}"
    
    geojson_point = {
        "type": "FeatureCollection",
        "features": [
            {
                "type": "Feature",
                "geometry": {
                    "type": "Point",
                    "coordinates": [lon, lat]
                },
                "properties": {
                    "candidato": candidato,
                    "local": nm_local,
                    "votos": f"{votos} ({pct}%)",
                    "votos_validos": f"{votos_validos}%"
                }
            }
        ]
    }
    
    tooltip_customizado = folium.GeoJsonTooltip(
        fields=['local'],
        aliases=['Local:'],
        style=estilo_tooltip,
        sticky=True,
        localize=True
    )
    
    marker = folium.Circle(
        radius=math.sqrt(row.QT_VOTOS) * 5,
        color=color if not vencedor else '#fca503',
        fill=True,
        fill_color=color,
        fill_opacity=min(max(0.1, row.DOMINANCIA_LOCAL), 0.80)
    )
   
    popup = folium.Popup(
        html_popup,
        max_width=350
    )
    
    folium.GeoJson(
        geojson_point,
        marker=marker,
        popup=popup,
        tooltip=tooltip_customizado    
    ).add_to(camada)
    
# Adiciona as camadas ao mapa 
for camada in camadas_candidatos_2T.values():
    camada.add_to(mapa_2T_locais)
folium.LayerControl(collapsed=False).add_to(mapa_2T_locais)
```

**Justificativa escolhas gráficas**

Decidimos por utilizar, para cada candidato, a cor dominante do partido/campanha, para que facilite a associação do leitor da cor com o candidato em questão.

Também decidimos por utilizar uma opacidade diferente em cada local para representar a força da votação do candidato naquele local. A dominância é calculada em relação com as demais votações obtidas no município. Utilizamos os percentis 10% e 90% como votações de referência para não sofrer o impacto causado por outliers.

Neste mapa, realizamos uma visualização em camadas para que o leitor possa comparar facilmente qual dos dois candidato foi o vencedor e quão grande foi a vantagem obtida. Também é possível que o leitor compare o impacto dos votos brancos/nulos em cada uma das seções, se ele se interessar.

Quanto à escala, escolhemos uma que fosse capaz de mostrar todo o mapa de São João del-Rei já a primeira vista, para que o panorama prinicipal já ficasse a disposição do leitor a primeira vista. 

Por fim, utilizamos uma borda amarela/laranja para marcar que o candidato foi aquele com maior votação naquele local.
